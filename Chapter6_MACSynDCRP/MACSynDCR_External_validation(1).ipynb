```python
# @title
import torch
import numpy as np
from rdkit import Chem
from rdkit.Chem import rdmolops

def atom_features(atom) -> torch.Tensor:
    """Extract a binary/numerical feature vector for one atom."""
    features = [
        atom.GetDegree(),                                # Num neighboring atoms
        atom.GetTotalNumHs(),                            # Num hydrogen atoms
        atom.GetTotalValence(),              # Valence
        int(atom.GetIsotope() > 0),                      # Isotope
        int(atom.GetIsAromatic())                        # Aromaticity
    ]
    return torch.tensor(features, dtype=torch.float32)

def build_molecular_graph(smiles: str):
    """Construct graph and features from SMILES using RDKit."""
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol)
    n_atoms = mol.GetNumAtoms()

    A = torch.zeros((n_atoms, n_atoms))
    atom_feats = []

    for atom in mol.GetAtoms():
        atom_feats.append(atom_features(atom))

    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        A[i, j] = 1
        A[j, i] = 1

    atom_feats = torch.stack(atom_feats)  # Shape: (n_atoms, q)
    return A, atom_feats

def laplacian_positional_encoding(A: torch.Tensor, k: int = None) -> torch.Tensor:
    """Compute Laplacian eigenvectors for position encoding."""
    n = A.shape[0]
    D = torch.diag(A.sum(dim=1))
    D_inv_sqrt = torch.linalg.inv(torch.sqrt(D))
    L = torch.eye(n) - D_inv_sqrt @ A @ D_inv_sqrt

    eigvals, eigvecs = torch.linalg.eigh(L)  # Sorted ascending
    return eigvecs  # Shape: (n, n)

class PositionalEncoder(torch.nn.Module):
    """Linear layer for projecting positional encoding."""
    def __init__(self, input_dim: int, output_dim: int):
        super().__init__()
        self.linear = torch.nn.Linear(input_dim, output_dim)

    def forward(self, U: torch.Tensor) -> torch.Tensor:
        return self.linear(U)

def prepare_drug_input(smiles: str, feature_dim: int = 5, pos_dim: int = 5) -> torch.Tensor:
    A, z = build_molecular_graph(smiles)  # Adjacency matrix and atom features
    U = laplacian_positional_encoding(A)  # Positional encodings

    linear = PositionalEncoder(input_dim=U.shape[1], output_dim=feature_dim)
    U_proj = linear(U)  # Shape: (n_atoms, q)

    X0 = z + U_proj  # x(0)_v = z_v + Linear(U_v)
    return X0  # Shape: (n_atoms, q)

# 🔬 Example
if __name__ == "__main__":
    smiles = "CC(=O)Oc1ccccc1C(=O)O"  # Aspirin
    X_0 = prepare_drug_input(smiles)
    print("X(0) shape:", X_0.shape)
    print("X(0) matrix:\n", X_0)
```

```python
# @title
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import pandas as pd
from rdkit import Chem
import random
from tqdm import tqdm
import argparse

# ---------- Feature Functions ----------
def atom_features(atom):
    return torch.tensor([
        atom.GetDegree(),
        atom.GetTotalNumHs(),
        atom.GetTotalValence(),
        int(atom.GetIsotope() > 0),
        int(atom.GetIsAromatic())
    ], dtype=torch.float32)

def bond_features(bond):
    return torch.tensor([
        int(bond.GetBondTypeAsDouble()),
        int(bond.GetIsConjugated())
    ], dtype=torch.float32)

def smiles_to_graph(smiles):
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol)
    n = mol.GetNumAtoms()

    node_feats = []
    edge_index = []
    edge_feats = []

    for atom in mol.GetAtoms():
        node_feats.append(atom_features(atom))

    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        edge_index.append([i, j])
        edge_index.append([j, i])
        edge_feats.append(bond_features(bond))
        edge_feats.append(bond_features(bond))

    node_feats = torch.stack(node_feats)
    edge_index = torch.tensor(edge_index).T
    edge_feats = torch.stack(edge_feats)

    adj = torch.zeros((n, n))
    for i, j in edge_index.T:
        adj[i, j] = 1

    return node_feats, edge_index, edge_feats, adj

def laplacian_positional_encoding(adj, k=32):
    n = adj.size(0)
    D = torch.diag(adj.sum(dim=1))
    D_inv_sqrt = torch.linalg.inv(torch.sqrt(D))
    L = torch.eye(n) - D_inv_sqrt @ adj @ D_inv_sqrt
    _, eigvecs = torch.linalg.eigh(L)

    if n >= k:
        return eigvecs[:, :k]
    else:
        padding = torch.zeros(n, k - n, device=eigvecs.device)
        return torch.cat([eigvecs, padding], dim=1)

# ---------- MGTN and GRAN Layers ----------
class GRANLayer(nn.Module):
    def __init__(self, in_dim, edge_dim, hidden_dim, num_samples=3):
        super().__init__()
        self.num_samples = num_samples
        self.hidden_dim = hidden_dim
        self.edge_dim = edge_dim
        self.attn_fc = nn.Linear(2 * in_dim + edge_dim, hidden_dim)
        self.lambda_param = nn.Parameter(torch.empty(hidden_dim))
        self.message_mlp = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, in_dim)
        )
        nn.init.xavier_uniform_(self.lambda_param.view(1, -1))

    def forward(self, x, edge_index, edge_attr):
        n = x.size(0)
        device = x.device
        adj_list = [[] for _ in range(n)]
        edge_dict = {}

        for idx in range(edge_index.shape[1]):
            i, j = edge_index[:, idx]
            adj_list[i.item()].append(j.item())
            edge_dict[(i.item(), j.item())] = edge_attr[idx]

        x_gran = torch.zeros_like(x)

        for v in range(n):
            sampled = random.sample(adj_list[v], min(len(adj_list[v]), self.num_samples)) if adj_list[v] else []
            agg_messages = []
            attn_scores = []

            for u in sampled:
                e = edge_dict.get((v, u), torch.zeros(self.edge_dim, device=device))
                h = torch.cat([x[v], e, x[u]], dim=0)
                phi = F.leaky_relu(self.attn_fc(h))
                alpha = torch.dot(self.lambda_param, phi)
                attn_scores.append(alpha)
                agg_messages.append(self.message_mlp(F.relu(x[u])))

            if attn_scores:
                weights = F.softmax(torch.stack(attn_scores), dim=0)
                weighted_sum = sum(w * m for w, m in zip(weights, agg_messages))
                x_gran[v] = weighted_sum + x[v]

        return x_gran

class MGTNLayer(nn.Module):
    def __init__(self, in_dim):
        super().__init__()
        self.q_fc = nn.Linear(in_dim, in_dim)
        self.k_fc = nn.Linear(in_dim, in_dim)
        self.v_fc = nn.Linear(in_dim, in_dim)
        self.combine = nn.Linear(in_dim, in_dim)

    def forward(self, x):
        Q = self.q_fc(x)
        K = self.k_fc(x)
        V = self.v_fc(x)
        scores = torch.matmul(Q, K.T) / (Q.shape[-1] ** 0.5)
        attn = F.softmax(scores, dim=-1)
        context = torch.matmul(attn, V)
        return self.combine(context)

# ---------- Encoder ----------
class MGTNSynDrugEncoder(nn.Module):
    def __init__(self, feat_dim, pos_dim, edge_dim, hidden_dim):
        super().__init__()
        self.atom_fc = nn.Linear(5, feat_dim)
        self.linear_pos = nn.Linear(pos_dim, feat_dim)
        self.gran = GRANLayer(in_dim=feat_dim, edge_dim=edge_dim, hidden_dim=hidden_dim)
        self.mgtn = MGTNLayer(in_dim=feat_dim)
        self.encoder = nn.Linear(feat_dim * 2, feat_dim)

    def forward(self, smiles):
        x, edge_index, edge_attr, adj = smiles_to_graph(smiles)
        x = self.atom_fc(x)
        U = laplacian_positional_encoding(adj, k=32)
        U_proj = self.linear_pos(U)
        x0 = x + U_proj
        x_gran = self.gran(x0, edge_index, edge_attr)
        x_mgtn = self.mgtn(x0)
        x_comb = torch.cat([x_gran, x_mgtn], dim=-1)
        return F.relu(self.encoder(x_comb))

# ---------- VAE ----------
class DrugComboVAE(nn.Module):
    def __init__(self, feat_dim=64, latent_dim=128, pos_dim=32, edge_dim=2, hidden_dim=128):
        super().__init__()
        self.encoder = MGTNSynDrugEncoder(feat_dim, pos_dim, edge_dim, hidden_dim)
        self.fc_mu = nn.Linear(feat_dim, latent_dim)
        self.fc_logvar = nn.Linear(feat_dim, latent_dim)
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.ReLU(),
            nn.Linear(128, feat_dim)
        )
        self.classifier = nn.Sequential(
            nn.Linear(latent_dim * 2, 128),
            nn.ReLU(),
            nn.Linear(128, 1)
        )

    def encode(self, smiles):
        x = self.encoder(smiles)
        h = x.mean(dim=0)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        z = mu + eps * std
        return z, mu, logvar, h

    def forward(self, s1, s2):
        z1, mu1, lv1, h1 = self.encode(s1)
        z2, mu2, lv2, h2 = self.encode(s2)
        r1 = self.decoder(z1)
        r2 = self.decoder(z2)
        out = self.classifier(torch.cat([z1, z2], dim=-1))
        return out, (r1, h1), (r2, h2), (mu1, lv1), (mu2, lv2)

# ---------- Dataset ----------
class DrugComboDataset(Dataset):
    def __init__(self, csv_path):
        df = pd.read_csv(csv_path)
        self.smiles1 = df['drug_a_name'].tolist()
        self.smiles2 = df['drug_b_name'].tolist()
        self.labels = df['SynClass'].tolist()

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return self.smiles1[idx], self.smiles2[idx], torch.tensor(self.labels[idx], dtype=torch.float32)

# ---------- Loss ----------
def combo_vae_loss(pred, label, r1, h1, r2, h2, mu1, lv1, mu2, lv2):
    pred_loss = F.mse_loss(pred.view(-1), label.view(-1))
    recon_loss = F.mse_loss(r1, h1) + F.mse_loss(r2, h2)
    kl = -0.5 * torch.sum(1 + lv1 - mu1.pow(2) - lv1.exp()) - 0.5 * torch.sum(1 + lv2 - mu2.pow(2) - lv2.exp())
    return pred_loss + recon_loss + 1e-3 * kl

# ---------- Training ----------
def train(args):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    temp_csv_path = 'temp_drug_combinations.csv'
    args.data.to_csv(temp_csv_path, index=False)
    dataset = DrugComboDataset(csv_path=temp_csv_path)
    loader = DataLoader(dataset, batch_size=1, shuffle=True)
    model = DrugComboVAE().to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=args.lr)

    for epoch in range(args.epochs):
        model.train()
        total_loss = 0
        for s1, s2, y in tqdm(loader):
            s1, s2, y = s1[0], s2[0], y.to(device)
            try:
                out, (r1, h1), (r2, h2), (mu1, lv1), (mu2, lv2) = model(s1, s2)
                loss = combo_vae_loss(out, y, r1, h1, r2, h2, mu1, lv1, mu2, lv2)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                total_loss += loss.item()
            except Exception as e:
                print(f"Error in combo: {e}")
        print(f"Epoch {epoch+1}, Loss: {total_loss / len(loader):.4f}")

```

```python
# @title
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import pandas as pd
from rdkit import Chem
import random
from tqdm import tqdm
import argparse
import os

# ---------- Feature Functions ----------
def atom_features(atom):
    return torch.tensor([
        atom.GetDegree(),
        atom.GetTotalNumHs(),
        atom.GetTotalValence(),
        int(atom.GetIsotope() > 0),
        int(atom.GetIsAromatic())
    ], dtype=torch.float32)

def bond_features(bond):
    return torch.tensor([
        int(bond.GetBondTypeAsDouble()),
        int(bond.GetIsConjugated())
    ], dtype=torch.float32)

def smiles_to_graph(smiles):
    mol = Chem.MolFromSmiles(smiles)
    # Handle cases where SMILES are invalid
    if mol is None:
        # Return empty tensors or handle as appropriate for your model
        return torch.empty(0, 5), torch.empty(2, 0, dtype=torch.long), torch.empty(0, 2), torch.empty(0, 0)

    mol = Chem.AddHs(mol)
    n = mol.GetNumAtoms()

    node_feats = []
    edge_index = []
    edge_feats = []

    for atom in mol.GetAtoms():
        node_feats.append(atom_features(atom))

    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        edge_index.append([i, j])
        edge_index.append([j, i])
        edge_feats.append(bond_features(bond))
        edge_feats.append(bond_features(bond))

    if not node_feats: # Handle empty list if no atoms
        node_feats = torch.empty(0, 5)
    else:
        node_feats = torch.stack(node_feats)

    if not edge_index: # Handle empty list if no bonds
        edge_index = torch.empty(2, 0, dtype=torch.long)
        edge_feats = torch.empty(0, 2)
        adj = torch.empty(n, n) # Initialize adjacency for a molecule with no bonds
    else:
        edge_index = torch.tensor(edge_index).T
        edge_feats = torch.stack(edge_feats)
        adj = torch.zeros((n, n))
        for i, j in edge_index.T:
            adj[i, j] = 1


    return node_feats, edge_index, edge_feats, adj


def laplacian_positional_encoding(adj, k=32):
    n = adj.size(0)
    if n == 0: # Handle empty adjacency matrix
        return torch.zeros(0, k, device=adj.device) # Ensure device consistency
    D = torch.diag(adj.sum(dim=1))
    # Handle potential division by zero if a node has no neighbors by adding epsilon
    D_inv_sqrt = torch.linalg.inv(torch.sqrt(D + 1e-8)) # Add small epsilon for stability
    L = torch.eye(n, device=adj.device) - D_inv_sqrt @ adj @ D_inv_sqrt

    # Ensure L is symmetric before eigh
    L = (L + L.T) * 0.5

    try:
        eigvals, eigvecs = torch.linalg.eigh(L)
    except Exception as e:
        print(f"Error during eigendecomposition: {e}")
        # Return a zero tensor of expected size in case of an error
        return torch.zeros(n, k, device=adj.device)


    if n >= k:
        return eigvecs[:, :k]
    else:
        padding = torch.zeros(n, k - n, device=eigvecs.device)
        return torch.cat([eigvecs, padding], dim=1)

# ---------- MGTN and GRAN Layers ----------
class GRANLayer(nn.Module):
    def __init__(self, in_dim, edge_dim, hidden_dim, num_samples=3):
        super().__init__()
        self.num_samples = num_samples
        self.hidden_dim = hidden_dim
        self.edge_dim = edge_dim
        self.attn_fc = nn.Linear(2 * in_dim + edge_dim, hidden_dim)
        self.lambda_param = nn.Parameter(torch.empty(hidden_dim))
        self.message_mlp = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, in_dim)
        )
        nn.init.xavier_uniform_(self.lambda_param.view(1, -1))

    def forward(self, x, edge_index, edge_attr):
        n = x.size(0)
        if n == 0: # Handle empty input
            return torch.empty_like(x)

        device = x.device
        adj_list = [[] for _ in range(n)]
        edge_dict = {}

        # Check if edge_index is empty
        if edge_index.numel() > 0:
            for idx in range(edge_index.shape[1]):
                i, j = edge_index[:, idx]
                adj_list[i.item()].append(j.item())
                edge_dict[(i.item(), j.item())] = edge_attr[idx]

        x_gran = torch.zeros_like(x)

        for v in range(n):
            sampled = random.sample(adj_list[v], min(len(adj_list[v]), self.num_samples)) if adj_list[v] else []
            agg_messages = []
            attn_scores = []

            for u in sampled:
                e = edge_dict.get((v, u), torch.zeros(self.edge_dim, device=device))
                h = torch.cat([x[v], e, x[u]], dim=0)
                phi = F.leaky_relu(self.attn_fc(h))
                alpha = torch.dot(self.lambda_param, phi)
                attn_scores.append(alpha)
                agg_messages.append(self.message_mlp(F.relu(x[u])))

            if attn_scores:
                weights = F.softmax(torch.stack(attn_scores), dim=0)
                weighted_sum = sum(w * m for w, m in zip(weights, agg_messages))
                x_gran[v] = weighted_sum + x[v]
            else:
                # If no neighbors were sampled (e.g., isolated node), the message is zero
                 x_gran[v] = x[v]


        return x_gran

class MGTNLayer(nn.Module):
    def __init__(self, in_dim):
        super().__init__()
        self.q_fc = nn.Linear(in_dim, in_dim)
        self.k_fc = nn.Linear(in_dim, in_dim)
        self.v_fc = nn.Linear(in_dim, in_dim)
        self.combine = nn.Linear(in_dim, in_dim)

    def forward(self, x):
        if x.size(0) == 0: # Handle empty input
            return torch.empty_like(x)
        Q = self.q_fc(x)
        K = self.k_fc(x)
        V = self.v_fc(x)
        scores = torch.matmul(Q, K.T) / (Q.shape[-1] ** 0.5)
        attn = F.softmax(scores, dim=-1)
        context = torch.matmul(attn, V)
        return self.combine(context)

# ---------- Encoder ----------
class MGTNSynDrugEncoder(nn.Module):
    def __init__(self, feat_dim, pos_dim, edge_dim, hidden_dim):
        super().__init__()
        self.atom_fc = nn.Linear(5, feat_dim)
        self.linear_pos = nn.Linear(pos_dim, feat_dim)
        self.gran = GRANLayer(in_dim=feat_dim, edge_dim=edge_dim, hidden_dim=hidden_dim)
        self.mgtn = MGTNLayer(in_dim=feat_dim)
        self.encoder = nn.Linear(feat_dim * 2, feat_dim)

    def forward(self, smiles):
        x, edge_index, edge_attr, adj = smiles_to_graph(smiles)
        if x.size(0) == 0: # Handle cases where smiles_to_graph returns empty tensors
            # Return a zero tensor of appropriate size for the rest of the pipeline
            return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)

        x = self.atom_fc(x)
        U = laplacian_positional_encoding(adj, k=32)

        # Ensure U has the correct number of dimensions (n_atoms, pos_dim)
        # If smiles_to_graph returns empty, U might be (0, k), need to handle proj
        if U.size(0) == 0:
             return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)

        U_proj = self.linear_pos(U)

        # Ensure x0, x_gran, x_mgtn are not empty before concatenation
        if x.size(0) == 0:
            return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)

        x0 = x + U_proj

        x_gran = self.gran(x0, edge_index, edge_attr)
        x_mgtn = self.mgtn(x0)

        # Ensure the concatenated tensor is not empty
        if x_gran.size(0) == 0 or x_mgtn.size(0) == 0:
             return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)


        x_comb = torch.cat([x_gran, x_mgtn], dim=-1)
        # Aggregate node features to get a single graph-level feature vector
        # Ensure x_comb is not empty before calling mean
        if x_comb.size(0) == 0:
             return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)


        graph_feat = F.relu(self.encoder(x_comb)).mean(dim=0)
        return graph_feat

# ---------- VAE ----------
class DrugComboVAE(nn.Module):
    def __init__(self, feat_dim=64, latent_dim=128, pos_dim=32, edge_dim=2, hidden_dim=128):
        super().__init__()
        # Encoder is now a graph-level encoder
        self.encoder = MGTNSynDrugEncoder(feat_dim, pos_dim, edge_dim, hidden_dim)
        self.fc_mu = nn.Linear(feat_dim, latent_dim)
        self.fc_logvar = nn.Linear(feat_dim, latent_dim)
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.ReLU(),
            # The decoder output needs to match the input to the encoder's final aggregation
            # Since we are aggregating to a single vector, the decoder reconstructs this vector
            nn.Linear(128, feat_dim)
        )
        self.classifier = nn.Sequential(
            nn.Linear(latent_dim * 2, 128),
            nn.ReLU(),
            nn.Linear(128, 1)
        )

    def encode(self, smiles):
        # The encoder directly returns a graph-level feature vector
        h = self.encoder(smiles)
        # Ensure h has a batch dimension for linear layers if needed
        if h.ndim == 1:
            h = h.unsqueeze(0)

        # Handle cases where the encoder returns a zero tensor (e.g., for invalid SMILES)
        if h.size(0) == 0:
             # Return zero tensors of appropriate size
             mu = torch.zeros(1, self.fc_mu.out_features, device=self.fc_mu.weight.device)
             logvar = torch.zeros(1, self.fc_logvar.out_features, device=self.fc_logvar.weight.device)
             z = torch.zeros(1, self.fc_mu.out_features, device=self.fc_mu.weight.device)
             return z, mu, logvar, torch.zeros(1, self.encoder.encoder.out_features, device=self.encoder.encoder.weight.device)


        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)

        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        z = mu + eps * std
        return z, mu, logvar, h # h is the graph-level feature before mu/logvar layers

    def forward(self, s1, s2):
        z1, mu1, lv1, h1 = self.encode(s1)
        z2, mu2, lv2, h2 = self.encode(s2)
        # The decoder reconstructs the graph-level feature vectors h1 and h2
        r1 = self.decoder(z1)
        r2 = self.decoder(z2)

        # Ensure z1 and z2 have the same size for concatenation, handling empty graphs
        if z1.size(0) == 0 and z2.size(0) == 0:
             out = torch.zeros(1, 1, device=self.classifier[0].weight.device)
        elif z1.size(0) == 0:
             z1 = torch.zeros(1, self.fc_mu.out_features, device=z2.device)
             out = self.classifier(torch.cat([z1, z2], dim=-1))
        elif z2.size(0) == 0:
             z2 = torch.zeros(1, self.fc_mu.out_features, device=z1.device)
             out = self.classifier(torch.cat([z1, z2], dim=-1))
        else:
            out = self.classifier(torch.cat([z1, z2], dim=-1))


        return out, (r1, h1), (r2, h2), (mu1, lv1), (mu2, lv2)

# ---------- Dataset ----------
class DrugComboDataset(Dataset):
    def __init__(self, csv_path):
        df = pd.read_csv(csv_path)
        # Ensure the column names are correct
        self.smiles1 = df['drug1'].tolist()
        self.smiles2 = df['drug2'].tolist()
        self.labels = df['label'].tolist()

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return self.smiles1[idx], self.smiles2[idx], torch.tensor(self.labels[idx], dtype=torch.float32)

# ---------- Loss ----------
def combo_vae_loss(pred, label, r1, h1, r2, h2, mu1, lv1, mu2, lv2):
    # Ensure all tensors have the same number of dimensions for F.mse_loss
    # Add batch dimension to single-sample tensors if needed by mse_loss or other operations
    if pred.ndim == 0: pred = pred.unsqueeze(0)
    if label.ndim == 0: label = label.unsqueeze(0)
    if r1.ndim == 1: r1 = r1.unsqueeze(0)
    if h1.ndim == 1: h1 = h1.unsqueeze(0)
    if r2.ndim == 1: r2 = r2.unsqueeze(0)
    if h2.ndim == 1: h2 = h2.unsqueeze(0)
    if mu1.ndim == 1: mu1 = mu1.unsqueeze(0)
    if lv1.ndim == 1: lv1 = lv1.unsqueeze(0)
    if mu2.ndim == 1: mu2 = mu2.unsqueeze(0)
    if lv2.ndim == 1: lv2 = lv2.unsqueeze(0)


    pred_loss = F.mse_loss(pred.view(-1), label.view(-1))

    # Handle cases where h1 or h2 might be zero tensors from empty graphs
    recon_loss = 0
    if h1.numel() > 0:
        recon_loss += F.mse_loss(r1, h1)
    if h2.numel() > 0:
        recon_loss += F.mse_loss(r2, h2)

    # Handle cases where mu or logvar might be zero tensors
    kl = 0
    if mu1.numel() > 0 and lv1.numel() > 0:
         kl += -0.5 * torch.sum(1 + lv1 - mu1.pow(2) - lv1.exp())
    if mu2.numel() > 0 and lv2.numel() > 0:
         kl += -0.5 * torch.sum(1 + lv2 - mu2.pow(2) - lv2.exp())


    return pred_loss + recon_loss + 1e-3 * kl
```

```python
# @title
# ---------- Training ----------
def train(args, model):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    temp_csv_dir = '.' # Use current directory for temp files
    temp_csv_path = os.path.join(temp_csv_dir, 'temp_drug_combinations.csv')

    # Rename columns to match expected dataset format
    renamed_data = args.data.rename(columns={'drug_a_name': 'drug1', 'drug_b_name': 'drug2', 'SynClass': 'label'})
    renamed_data.to_csv(temp_csv_path, index=False)

    dataset = DrugComboDataset(csv_path=temp_csv_path)
    # Use batch_size 1 as the smiles_to_graph handles one SMILES string at a time
    loader = DataLoader(dataset, batch_size=1, shuffle=True)
    model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=args.lr)

    for epoch in range(args.epochs):
        model.train()
        total_loss = 0
        # Added enumerate to get batch_idx for potential debugging or logging
        for batch_idx, (s1, s2, y) in enumerate(tqdm(loader)):
            # Dataloader with batch_size=1 will return tensors of shape (1, ...)
            # Extract the single item from the batch
            s1, s2, y = s1[0], s2[0], y.to(device)

            try:
                out, (r1, h1), (r2, h2), (mu1, lv1), (mu2, lv2) = model(s1, s2)
                # Ensure out, y, r1, h1, r2, h2, mu1, lv1, mu2, lv2 are correctly shaped for the loss function
                # Squeeze any extra batch dimensions if batch_size is 1
                loss = combo_vae_loss(out.squeeze(0).squeeze(-1), y.squeeze(0), r1.squeeze(0), h1.squeeze(0), r2.squeeze(0), h2.squeeze(0), mu1.squeeze(0), lv1.squeeze(0), mu2.squeeze(0), lv2.squeeze(0))
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                total_loss += loss.item()
            except Exception as e:
                print(f"Error processing combo (SMILES: {s1}, {s2}) in batch {batch_idx}: {e}")
                # Optionally, you can skip the batch or log the error
                continue # Skip to the next batch

        # Calculate average loss per sample or per batch
        avg_loss = total_loss / len(loader) if len(loader) > 0 else 0
        print(f"Epoch {epoch+1}, Loss: {avg_loss:.4f}")

    # Clean up the temporary CSV file
    if os.path.exists(temp_csv_path):
        os.remove(temp_csv_path)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import pandas as pd
from rdkit import Chem
import random
from tqdm import tqdm
import argparse
import os

# ---------- Feature Functions ----------
def atom_features(atom):
    return torch.tensor([
        atom.GetDegree(),
        atom.GetTotalNumHs(),
        atom.GetTotalValence(),
        int(atom.GetIsotope() > 0),
        int(atom.GetIsAromatic())
    ], dtype=torch.float32)

def bond_features(bond):
    return torch.tensor([
        int(bond.GetBondTypeAsDouble()),
        int(bond.GetIsConjugated())
    ], dtype=torch.float32)

def smiles_to_graph(smiles):
    mol = Chem.MolFromSmiles(smiles)
    # Handle cases where SMILES are invalid
    if mol is None:
        # Return empty tensors or handle as appropriate for your model
        return torch.empty(0, 5), torch.empty(2, 0, dtype=torch.long), torch.empty(0, 2), torch.empty(0, 0)

    mol = Chem.AddHs(mol)
    n = mol.GetNumAtoms()

    node_feats = []
    edge_index = []
    edge_feats = []

    for atom in mol.GetAtoms():
        node_feats.append(atom_features(atom))

    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        edge_index.append([i, j])
        edge_index.append([j, i])
        edge_feats.append(bond_features(bond))
        edge_feats.append(bond_features(bond))

    if not node_feats: # Handle empty list if no atoms
        node_feats = torch.empty(0, 5)
    else:
        node_feats = torch.stack(node_feats)

    if not edge_index: # Handle empty list if no bonds
        edge_index = torch.empty(2, 0, dtype=torch.long)
        edge_feats = torch.empty(0, 2)
        adj = torch.empty(n, n) # Initialize adjacency for a molecule with no bonds
    else:
        edge_index = torch.tensor(edge_index).T
        edge_feats = torch.stack(edge_feats)
        adj = torch.zeros((n, n))
        for i, j in edge_index.T:
            adj[i, j] = 1


    return node_feats, edge_index, edge_feats, adj


def laplacian_positional_encoding(adj, k=64):
    n = adj.size(0)
    if n == 0: # Handle empty adjacency matrix
        return torch.zeros(0, k, device=adj.device) # Ensure device consistency
    D = torch.diag(adj.sum(dim=1))
    # Handle potential division by zero if a node has no neighbors by adding epsilon
    D_inv_sqrt = torch.linalg.inv(torch.sqrt(D + 1e-8)) # Add small epsilon for stability
    L = torch.eye(n, device=adj.device) - D_inv_sqrt @ adj @ D_inv_sqrt

    # Ensure L is symmetric before eigh
    L = (L + L.T) * 0.5

    try:
        eigvals, eigvecs = torch.linalg.eigh(L)
    except Exception as e:
        print(f"Error during eigendecomposition: {e}")
        # Return a zero tensor of expected size in case of an error
        return torch.zeros(n, k, device=adj.device)


    if n >= k:
        return eigvecs[:, :k]
    else:
        padding = torch.zeros(n, k - n, device=eigvecs.device)
        return torch.cat([eigvecs, padding], dim=1)

# ---------- MGTN and GRAN Layers ----------
class GRANLayer(nn.Module):
    def __init__(self, in_dim, edge_dim, hidden_dim, num_samples=3):
        super().__init__()
        self.num_samples = num_samples
        self.hidden_dim = hidden_dim
        self.edge_dim = edge_dim
        self.attn_fc = nn.Linear(2 * in_dim + edge_dim, hidden_dim)
        self.lambda_param = nn.Parameter(torch.empty(hidden_dim))
        self.message_mlp = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, in_dim)
        )
        nn.init.xavier_uniform_(self.lambda_param.view(1, -1))

    def forward(self, x, edge_index, edge_attr):
        n = x.size(0)
        if n == 0: # Handle empty input
            return torch.empty_like(x)

        device = x.device
        adj_list = [[] for _ in range(n)]
        edge_dict = {}

        # Check if edge_index is empty
        if edge_index.numel() > 0:
            for idx in range(edge_index.shape[1]):
                i, j = edge_index[:, idx]
                adj_list[i.item()].append(j.item())
                edge_dict[(i.item(), j.item())] = edge_attr[idx]

        x_gran = torch.zeros_like(x)

        for v in range(n):
            sampled = random.sample(adj_list[v], min(len(adj_list[v]), self.num_samples)) if adj_list[v] else []
            agg_messages = []
            attn_scores = []

            for u in sampled:
                e = edge_dict.get((v, u), torch.zeros(self.edge_dim, device=device))
                h = torch.cat([x[v], e, x[u]], dim=0)
                phi = F.leaky_relu(self.attn_fc(h))
                alpha = torch.dot(self.lambda_param, phi)
                attn_scores.append(alpha)
                agg_messages.append(self.message_mlp(F.relu(x[u])))

            if attn_scores:
                weights = F.softmax(torch.stack(attn_scores), dim=0)
                weighted_sum = sum(w * m for w, m in zip(weights, agg_messages))
                x_gran[v] = weighted_sum + x[v]
            else:
                # If no neighbors were sampled (e.g., isolated node), the message is zero
                 x_gran[v] = x[v]
        return x_gran

class MGTNLayer(nn.Module):
    def __init__(self, in_dim):
        super().__init__()
        self.q_fc = nn.Linear(in_dim, in_dim)
        self.k_fc = nn.Linear(in_dim, in_dim)
        self.v_fc = nn.Linear(in_dim, in_dim)
        self.combine = nn.Linear(in_dim, in_dim)

    def forward(self, x):
        if x.size(0) == 0: # Handle empty input
            return torch.empty_like(x)
        Q = self.q_fc(x)
        K = self.k_fc(x)
        V = self.v_fc(x)
        scores = torch.matmul(Q, K.T) / (Q.shape[-1] ** 0.5)
        attn = F.softmax(scores, dim=-1)
        context = torch.matmul(attn, V)
        return self.combine(context)

# ---------- Encoder ----------
class MGTNSynDrugEncoder(nn.Module):
    def __init__(self, feat_dim, pos_dim, edge_dim, hidden_dim):
        super().__init__()
        self.atom_fc = nn.Linear(5, feat_dim)
        self.linear_pos = nn.Linear(pos_dim, feat_dim)
        self.gran = GRANLayer(in_dim=feat_dim, edge_dim=edge_dim, hidden_dim=hidden_dim)
        self.mgtn = MGTNLayer(in_dim=feat_dim)
        self.encoder = nn.Linear(feat_dim * 2, feat_dim)

    def forward(self, smiles):
        x, edge_index, edge_attr, adj = smiles_to_graph(smiles)
        if x.size(0) == 0: # Handle cases where smiles_to_graph returns empty tensors
            # Return a zero tensor of appropriate size for the rest of the pipeline
            return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)

        x = self.atom_fc(x)
        U = laplacian_positional_encoding(adj, k=64)

        # Ensure U has the correct number of dimensions (n_atoms, pos_dim)
        # If smiles_to_graph returns empty, U might be (0, k), need to handle proj
        if U.size(0) == 0:
             return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)

        U_proj = self.linear_pos(U)

        # Ensure x0, x_gran, x_mgtn are not empty before concatenation
        if x.size(0) == 0:
            return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)

        x0 = x + U_proj

        x_gran = self.gran(x0, edge_index, edge_attr)
        x_mgtn = self.mgtn(x0)

        # Ensure the concatenated tensor is not empty
        if x_gran.size(0) == 0 or x_mgtn.size(0) == 0:
             return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)


        x_comb = torch.cat([x_gran, x_mgtn], dim=-1)
        # Aggregate node features to get a single graph-level feature vector
        # Ensure x_comb is not empty before calling mean
        if x_comb.size(0) == 0:
             return torch.zeros(self.encoder.out_features, device=self.atom_fc.weight.device)


        graph_feat = F.relu(self.encoder(x_comb)).mean(dim=0)
        return graph_feat
```

```python
# Define dimensions used by the encoder backbone
feat_dim = 64
pos_dim = 64
edge_dim = 2
hidden_dim = 128

class DrugVAEEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        # Ensure input dimension matches the output of MGTNSynDrugEncoder's mean pooling
        self.fc1 = nn.Linear(feat_dim, 256)
        self.fc_mu = nn.Linear(256, 128)
        self.fc_logvar = nn.Linear(256, 128)

    def forward(self, x):
        h = F.relu(self.fc1(x))
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

def reparameterize(mu, logvar):
    std = torch.exp(0.5 * logvar)
    eps = torch.randn_like(std)
    return mu + eps * std

class DrugVAEDecoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(128, 256)
        # Decoder output should match the input to the VAEEncoder, which is the pooled graph feature
        self.fc2 = nn.Linear(256, feat_dim)

    def forward(self, z):
        h = F.relu(self.fc1(z))
        return self.fc2(h)

class DrugVAE(nn.Module):
    def __init__(self, feat_dim, pos_dim, edge_dim, hidden_dim): # Initialize with dimension parameters
        super().__init__()
        # The encoder backbone is an instance of MGTNSynDrugEncoder
        self.encoder_backbone = MGTNSynDrugEncoder(feat_dim, pos_dim, edge_dim, hidden_dim)
        self.vae_encoder = DrugVAEEncoder()
        self.decoder = DrugVAEDecoder()

    def forward(self, smiles):
        # The encoder_backbone directly returns the graph-level feature vector
        X_pool = self.encoder_backbone(smiles)       # [feat_dim]

        # Handle cases where the encoder_backbone returns a zero tensor (e.g., for invalid SMILES)
        if X_pool.numel() == 0:
             # Return zero tensors of appropriate size
             mu = torch.zeros(1, self.vae_encoder.fc_mu.out_features, device=X_pool.device)
             logvar = torch.zeros(1, self.vae_encoder.fc_logvar.out_features, device=X_pool.device)
             z = torch.zeros(1, self.vae_encoder.fc_mu.out_features, device=X_pool.device)
             X_recon = torch.zeros(1, self.decoder.fc2.out_features, device=X_pool.device)
             return z.squeeze(0), mu.squeeze(0), logvar.squeeze(0), X_recon.squeeze(0) # Remove batch dim as it's single sample

        # Ensure X_pool has a batch dimension for the linear layers if needed
        if X_pool.ndim == 1:
            X_pool = X_pool.unsqueeze(0)

        mu, logvar = self.vae_encoder(X_pool)   # [1, 128], [1, 128]
        z = reparameterize(mu, logvar)          # [1, 128]
        X_recon = self.decoder(z)               # [1, feat_dim] (reconstructed pooled embedding)

        # Remove the batch dimension before returning
        return z.squeeze(0), mu.squeeze(0), logvar.squeeze(0), X_recon.squeeze(0)


def loss_fn(X_recon, X_pool, mu, logvar):
    # Ensure all tensors have the same number of dimensions for F.mse_loss
    if X_recon.ndim == 1: X_recon = X_recon.unsqueeze(0)
    if X_pool.ndim == 1: X_pool = X_pool.unsqueeze(0)
    if mu.ndim == 1: mu = mu.unsqueeze(0)
    if logvar.ndim == 1: logvar = logvar.unsqueeze(0)

    recon_loss = F.mse_loss(X_recon, X_pool)

    # Handle cases where mu or logvar might be zero tensors
    kl = 0
    if mu.numel() > 0 and logvar.numel() > 0:
         kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())

    return recon_loss + 0.001 * kl

# Instantiate the DrugVAE model correctly

```

```python
"""
    A simple implementation of Gaussian MLP Encoder and Decoder
"""

class Encoder(nn.Module):

    def __init__(self, input_dim, hidden_dim, latent_dim):
        super(Encoder, self).__init__()

        self.FC_input = nn.Linear(input_dim, hidden_dim)
        self.FC_input2 = nn.Linear(hidden_dim, hidden_dim)
        self.FC_mean  = nn.Linear(hidden_dim, latent_dim)
        self.FC_var   = nn.Linear (hidden_dim, latent_dim)

        self.LeakyReLU = nn.LeakyReLU(0.2)

        self.training = True

    def forward(self, x):
        h_       = self.LeakyReLU(self.FC_input(x))
        h_       = self.LeakyReLU(self.FC_input2(h_))
        mean     = self.FC_mean(h_)
        log_var  = self.FC_var(h_)                     # encoder produces mean and log of variance
                                                       #             (i.e., parateters of simple tractable normal distribution "q"

        return mean, log_var

class Decoder(nn.Module):
    def __init__(self, latent_dim, hidden_dim, output_dim):
        super(Decoder, self).__init__()
        self.FC_hidden = nn.Linear(latent_dim, hidden_dim)
        self.FC_hidden2 = nn.Linear(hidden_dim, hidden_dim)
        self.FC_output = nn.Linear(hidden_dim, output_dim)

        self.LeakyReLU = nn.LeakyReLU(0.2)

    def forward(self, x):
        h     = self.LeakyReLU(self.FC_hidden(x))
        h     = self.LeakyReLU(self.FC_hidden2(h))

        x_hat = torch.sigmoid(self.FC_output(h))
        return x_hat

class VAE(nn.Module):
    def __init__(self, Encoder, Decoder):
        super(VAE, self).__init__()
        self.Encoder = Encoder
        self.Decoder = Decoder

    def reparameterization(self, mean, var):
        epsilon = torch.randn_like(var).to(DEVICE)        # sampling epsilon
        z = mean + var*epsilon                          # reparameterization trick
        return z


    def forward(self, x):
        mean, log_var = self.Encoder(x)
        z = self.reparameterization(mean, torch.exp(0.5 * log_var)) # takes exponential function (log var -> var)
        x_hat  = self.Decoder(z)

        return x_hat, mean, log_var
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

class VAE(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super(VAE, self).__init__()
        # Encoder
        self.fc1 = nn.Linear(input_dim, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3_mean = nn.Linear(256, latent_dim)  # For mean
        self.fc3_logvar = nn.Linear(256, latent_dim)  # For log variance

        # Decoder
        self.fc4 = nn.Linear(latent_dim, 256)
        self.fc5 = nn.Linear(256, 512)
        self.fc6 = nn.Linear(512, input_dim)

    def encode(self, x):
        h = F.relu(self.fc1(x))
        h = F.relu(self.fc2(h))
        return self.fc3_mean(h), self.fc3_logvar(h)

    def reparameterize(self, mu, log_var):
        std = torch.exp(0.5*log_var)
        eps = torch.randn_like(std)
        return mu + eps*std

    def decode(self, z):
        h = F.relu(self.fc4(z))
        h = F.relu(self.fc5(h))
        return torch.sigmoid(self.fc6(h))

    def forward(self, x):
        mu, log_var = self.encode(x)
        z = self.reparameterize(mu, log_var)
        return self.decode(z), mu, log_var

def loss_function(recon_x, x, mu, log_var):
        BCE = F.binary_cross_entropy(recon_x, x, reduction='sum')
        KLD = -0.5 * torch.sum(1 + log_var - mu.pow(2) - log_var.exp())
        return BCE + KLD
```

```python
# @title
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

class VAE(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super(VAE, self).__init__()
        # Encoder
        self.fc1 = nn.Linear(input_dim, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3_mean = nn.Linear(256, latent_dim)  # For mean
        self.fc3_logvar = nn.Linear(256, latent_dim)  # For log variance

        # Decoder
        self.fc4 = nn.Linear(latent_dim, 256)
        self.fc5 = nn.Linear(256, 512)
        self.fc6 = nn.Linear(512, input_dim)

    def encode(self, x):
        h = F.relu(self.fc1(x))
        h = F.relu(self.fc2(h))
        return self.fc3_mean(h), self.fc3_logvar(h)

    def reparameterize(self, mu, log_var):
        std = torch.exp(0.5*log_var)
        eps = torch.randn_like(std)
        return mu + eps*std

    def decode(self, z):
        h = F.relu(self.fc4(z))
        h = F.relu(self.fc5(h))
        return torch.sigmoid(self.fc6(h))

    def forward(self, x):
        mu, log_var = self.encode(x)
        z = self.reparameterize(mu, log_var)
        return self.decode(z), mu, log_var

def loss_function(recon_x, x, mu, log_var):
        BCE = F.binary_cross_entropy(recon_x, x, reduction='sum')
        KLD = -0.5 * torch.sum(1 + log_var - mu.pow(2) - log_var.exp())
        return BCE + KLD
```

```python
# @title
import numpy as np
from sklearn.preprocessing import normalize
def to01(array):
    a = array.min()
    # ignore the Runtime Warning
    with np.errstate(divide='ignore'):
        b = 1. /(array.max() - array.min())
    if not(np.isfinite(b)):
        b = 0
    return np.vectorize(lambda x: b * (x - a))(array)
```

```python
# @title
# Parameter settings
input_dim = 567
latent_dim = 128  # desired dimension of latent space
learning_rate = 1e-4
batch_size = 38
epochs = 10000
import numpy as np
from sklearn.preprocessing import normalize

drug_des_x_train = pd.read_csv('/content/drive/MyDrive/MACSynDCR/Data/Drug/drug_target.csv')
drug_des_x_train=drug_des_x_train.iloc[:,:]
drug_des_x_train=np.array(drug_des_x_train)

drug_des_x_train[np.isnan(drug_des_x_train)] = 0
normed_drug_des = to01(drug_des_x_train)
#normed_matrix = normalize(drug_sm_x_train, axis=1, norm='l1')
#result = drug_sm_x_train/ np.linalg.norm(drug_sm_x_train, axis=-1)[:, np.newaxis]

drug_sm_x_train_tensor = torch.tensor(normed_drug_des, dtype=torch.float32)
train_dataset = TensorDataset(drug_sm_x_train_tensor, drug_sm_x_train_tensor)  # input and target are the same
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

# Model, optimizer, and training
model = VAE(input_dim=input_dim, latent_dim=latent_dim)
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

model.train()
for epoch in range(epochs):
    overall_loss = 0
    for batch_idx, (data, _) in enumerate(train_loader):
        optimizer.zero_grad()
        recon_batch, mu, log_var = model(data)
        loss = loss_function(recon_batch, data, mu, log_var)
        loss.backward()
        overall_loss += loss.item()
        optimizer.step()


   # print(f"Epoch {epoch+1}, Loss: {overall_loss / len(train_loader.dataset)}")
    if (epoch + 1) % 500 == 0:
         print(f"Epoch {epoch+1}, Loss: {overall_loss / len(train_loader.dataset)}")
```

```python
import torch
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GATConv
from torch.nn import Linear, Sequential, ReLU, BatchNorm1d
from rdkit import Chem
import numpy as np

def smiles_to_graph(smiles):
    mol = Chem.MolFromSmiles(smiles)
    atoms = mol.GetAtoms()
    x = torch.tensor([atom.GetAtomicNum() for atom in atoms], dtype=torch.float).view(-1, 1)
    x = torch.nn.functional.one_hot(x.to(torch.int64).view(-1) - 1, num_classes=128).to(torch.float)

    edges = []
    for bond in mol.GetBonds():
        edges.append([bond.GetBeginAtomIdx(), bond.GetEndAtomIdx()])
        edges.append([bond.GetEndAtomIdx(), bond.GetBeginAtomIdx()])

    edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()

    return Data(x=x, edge_index=edge_index)
```

```python
import torch
import torch.nn.functional as F

def mol_resize(tensor,target_size):
    # Calculate padding
    padding_needed = target_size - tensor.size(0)
    # Pad only the first dimension (top and bottom)
    pad = (0, 0, padding_needed, 0)  # (left, right, top, bottom) for 2D
    # Apply padding
    padded_tensor = F.pad(tensor, pad, "constant", 0)  # Adding zeros
    return padded_tensor
```

```python
class DDIModel(torch.nn.Module):
    def __init__(self):
        super(DDIModel, self).__init__()
        self.conv1 = GATConv(in_channels=128, out_channels=32, heads=4, concat=True)
        self.conv2 = GATConv(in_channels=32*4, out_channels=128, heads=1, concat=False)
        self.fc = Sequential(Linear(2*128, 256), ReLU(), BatchNorm1d(256), Linear(256, 128))

    def forward(self, data1, data2):
        x1, edge_index1 = data1.x, data1.edge_index
        x2, edge_index2 = data2.x, data2.edge_index

        x1 = self.conv1(x1, edge_index1)
        x1 = torch.relu(x1)
        x1 = self.conv2(x1, edge_index1)

        x2 = self.conv1(x2, edge_index2)
        x2 = torch.relu(x2)
        x2 = self.conv2(x2, edge_index2)
        if x1.shape[0]>x2.shape[0]:
            x2=mol_resize(x2,x1.shape[0])
        elif x1.shape[0]<x2.shape[0]:
            x1=mol_resize(x1,x2.shape[0])

        # Concatenate the features from both drugs
        #x1 = torch.mean(x1, dim=0)
        #x2 = torch.mean(x2, dim=0)
        x = torch.cat([x1, x2], dim=1)
        x = self.fc(x)
        return x
```

```python
def extract_ddi_features(smiles1, smiles2, model):
    graph1 = smiles_to_graph(smiles1)
    graph2 = smiles_to_graph(smiles2)
    features = model(graph1, graph2)
    f=torch.mean(features,dim=0)
    return f.detach().numpy()
```

```python
# @title
import torch
from torch_geometric.data import Data
from torch_geometric.nn import GATConv
from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np

class GATNet(torch.nn.Module):
    def __init__(self):
        super(GATNet, self).__init__()
        self.conv1 = GATConv(in_channels=100, out_channels=128, heads=1)
        self.linear = torch.nn.Linear(128, 128)

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = self.conv1(x, edge_index)
        x = torch.sigmoid(x)
        x = torch.mean(x, dim=0)
        x = self.linear(x)
       ## print(x.shape)
        return x
```

```python
# @title
def smiles_to_graph(smiles):
    mol = Chem.MolFromSmiles(smiles)
    atoms = mol.GetAtoms()
    num_nodes = len(atoms)
    x = torch.tensor([atom.GetAtomicNum() for atom in atoms], dtype=torch.float).view(-1, 1)
    x = torch.nn.functional.one_hot(x.to(torch.int64).view(-1) - 1, num_classes=100).to(torch.float)

    edges = []
    for bond in mol.GetBonds():
        edges.append([bond.GetBeginAtomIdx(), bond.GetEndAtomIdx()])
        edges.append([bond.GetEndAtomIdx(), bond.GetBeginAtomIdx()])

    edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()

    data = Data(x=x, edge_index=edge_index)
    return data
```

```python
# @title
import numpy
def extract_features(smiles):
    model = GATNet()
    data = smiles_to_graph(smiles)
    features = model(data)
    return features.detach().numpy()

# Example usage:
DMG_features=[]
for sm in ext_smiles_list:
   # print(sm)
    features = extract_features(sm)
    #print(features.shape)  # Should be (128,)
    DMG_features.append(features)
DMG_features_128=numpy.array(DMG_features)
print(DMG_features_128.shape)
len(DMG_features)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler


# Example: X is NumPy array of shape (num_samples, 954)
ext_exprdata_numeric = ext_exprdata.apply(pd.to_numeric, errors='coerce')

# 2. Fill NaNs with 0 (or another strategy)
ext_exprdata_numeric = ext_exprdata_numeric.fillna(0)

# 3. Convert to NumPy array of float32
X_array = ext_exprdata_numeric.to_numpy(dtype=np.float32)

# 4. Convert to PyTorch tensor
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_array)  # mean=0, std=1

X_tensor = torch.tensor(X_scaled, dtype=torch.float32)

#X_tensor = torch.tensor(ext_exprdata, dtype=torch.float32)

dataset = TensorDataset(X_tensor)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

input_dim = 954
latent_dim = 256  # target dimension
hidden_dim = 768  # intermediate layer size

class VAE(nn.Module):
    def __init__(self, input_dim, hidden_dim, latent_dim):
        super(VAE, self).__init__()
        # Encoder
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc_mu = nn.Linear(hidden_dim, latent_dim)
        self.fc_logvar = nn.Linear(hidden_dim, latent_dim)
        # Decoder
        self.fc2 = nn.Linear(latent_dim, hidden_dim)
        self.fc3 = nn.Linear(hidden_dim, input_dim)
        self.relu = nn.ReLU()

    def encode(self, x):
        h = self.relu(self.fc1(x))
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        h = self.relu(self.fc2(z))
        return self.fc3(h)

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decode(z)
        return x_recon, mu, logvar
def vae_loss(recon_x, x, mu, logvar):
    recon_loss = nn.MSELoss()(recon_x, x)  # reconstruction
    kld = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return recon_loss + kld / x.size(0)
```

```python
vae = VAE(input_dim, hidden_dim, latent_dim)
optimizer = optim.Adam(vae.parameters(), lr=1e-3)
num_epochs = 500

vae.train()
for epoch in range(num_epochs):
    total_loss = 0
    for batch in dataloader:
        x_batch = batch[0]
        optimizer.zero_grad()
        recon_x, mu, logvar = vae(x_batch)
        loss = vae_loss(recon_x, x_batch, mu, logvar)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f"Epoch {epoch+1}, Loss: {total_loss/len(dataloader):.4f}")
```

```python
vae.eval()
with torch.no_grad():
    mu, logvar = vae.encode(X_tensor)
    X_reduced = mu.numpy()  # shape: (num_samples, 512)
print(X_reduced.shape)
X_reduced
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import (roc_auc_score, accuracy_score, average_precision_score,
                             balanced_accuracy_score, precision_score, recall_score,
                             f1_score, cohen_kappa_score, mean_squared_error)
from scipy.stats import pearsonr
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras import layers
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dropout, Dense
from tensorflow.keras.utils import to_categorical

# -------------------------------
# 1. Load data
# -------------------------------
train_df = pd.read_csv("/content/drive/MyDrive/MACSynDCR/Data/TranData_Ind_AstraZeneca_dataset_1664.csv")
test_df = pd.read_csv("/content/drive/MyDrive/MACSynDCR/Data/Exyernel_Ind_AstraZeneca_feature_dataset_1664.csv")

#X_train = train_df.iloc[:, :-1].values.astype(np.float32)
#y_train = train_df.iloc[:, -1].values.astype(int)

#X_test = test_df.iloc[:, :-1].values.astype(np.float32)
#y_test = test_df.iloc[:, -1].values.astype(int)

#X_train = train_df.iloc[:, :-1].to_numpy()  # all columns except last → features
#y_train = train_df.iloc[:, -1].to_numpy()   # last column → label

#X_test = test_df.iloc[:, :-1].to_numpy()
#y_test = test_df.iloc[:, -1].to_numpy()

# Assuming train_data and test_data are DataFrames
X_train = train_df.iloc[:, :1664].to_numpy(dtype=np.float32)  # all columns except last
print(X_train.shape)
y_train = train_df.iloc[:,1665].to_numpy(dtype=np.int64)
print(y_train)

X_test = test_df.iloc[:, :1664].to_numpy(dtype=np.float32)
print(X_test.shape)
y_test = test_df.iloc[:,1665].to_numpy(dtype=np.int64)

# Now reshape for CNN
# 1664 features → 52 x 32 x 1


# Standardize features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# -------------------------------
# 2. PyTorch ANN
# -------------------------------
class MACSynDCR_ANN(nn.Module):
    def __init__(self, input_dim):
        super(MACSynDCR_ANN, self).__init__()
        self.fc1 = nn.Linear(input_dim, 1024)
        self.bn1 = nn.BatchNorm1d(1024)
        self.fc3 = nn.Linear(1024, 512)
        self.bn3 = nn.BatchNorm1d(512)
        self.fc4 = nn.Linear(512, 256)
        self.bn4 = nn.BatchNorm1d(256)
        self.fc5 = nn.Linear(256, 64)
        self.bn5 = nn.BatchNorm1d(64)
        self.fc6 = nn.Linear(64, 2)

    def forward(self, x):
        x = F.relu(self.bn1(self.fc1(x)))
        x = F.relu(self.bn3(self.fc3(x)))
        x = F.relu(self.bn4(self.fc4(x)))
        x = F.relu(self.bn5(self.fc5(x)))
        x = F.softmax(self.fc6(x), dim=1)
        return x

def train_pytorch_ann(X_train, y_train, epochs=20, batch_size=32, lr=1e-3):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = MACSynDCR_ANN(X_train.shape[1]).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()

    dataset = TensorDataset(torch.tensor(X_train, dtype=torch.float32),
                            torch.tensor(y_train, dtype=torch.long))
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

    model.train()
    for epoch in range(epochs):
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            outputs = model(xb)
            loss = criterion(outputs, yb)
            loss.backward()
            optimizer.step()
    return model, device

def predict_pytorch_ann(model, device, X):
    model.eval()
    with torch.no_grad():
        X_tensor = torch.tensor(X, dtype=torch.float32).to(device)
        probs = model(X_tensor).cpu().numpy()[:,1]  # probability of class 1
    return probs

# -------------------------------
# 3. Keras ANN
# -------------------------------
def build_KANN_model(input_dim):
    model = Sequential([
        layers.Input(shape=(input_dim,)),
        layers.Dense(2048, activation='relu'),
        layers.BatchNormalization(),
        layers.Dense(1024, activation='relu'),
        layers.BatchNormalization(),
        layers.Dense(512, activation='relu'),
        layers.BatchNormalization(),
        layers.Dense(256, activation='relu'),
        layers.BatchNormalization(),
        layers.Dense(128, activation='relu'),
        layers.BatchNormalization(),
        layers.Dense(2, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

# -------------------------------
# 4. Keras CNN
# -------------------------------
def build_KCNN_model(input_shape):
    model = Sequential([
        Conv2D(32, (3,3), activation='relu', input_shape=input_shape),
        MaxPooling2D((2,2)),
        Dropout(0.2),
        Conv2D(64, (3,3), activation='relu'),
        MaxPooling2D((2,2)),
        Dropout(0.2),
        Conv2D(128, (3,3), activation='relu'),
        MaxPooling2D((2,2)),
        Dropout(0.2),
        Flatten(),
        Dense(128, activation='relu'),
        Dense(2, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

# Reshape data for CNN (example, adapt if needed)
#X_train_cnn = X_train.reshape(-1, 52, 32, 1)
#X_test_cnn = X_test.reshape(-1, 52, 32, 1)
X_train_cnn = X_train.reshape(-1, 52, 32, 1)
X_test_cnn  = X_test.reshape(-1, 52, 32, 1)
y_train_cat = to_categorical(y_train, 2)
y_test_cat = to_categorical(y_test, 2)

# -------------------------------
# 5. Train all models
# -------------------------------
# PyTorch ANN
ann_model, device = train_pytorch_ann(X_train, y_train, epochs=20)
ann_probs_test = predict_pytorch_ann(ann_model, device, X_test)

# Keras ANN
kann_model = build_KANN_model(X_train.shape[1])
kann_model.fit(X_train, y_train_cat, epochs=20, batch_size=32, verbose=0)
kann_probs_test = kann_model.predict(X_test)[:,1]

# Keras CNN
kcnn_model = build_KCNN_model((52,32,1))
kcnn_model.fit(X_train_cnn, y_train_cat, epochs=20, batch_size=32, verbose=0)
kcnn_probs_test = kcnn_model.predict(X_test_cnn)[:,1]

# -------------------------------
# 6. Stacking Ensemble with Logistic Regression
# -------------------------------
stacked_test = np.column_stack((ann_probs_test, kann_probs_test, kcnn_probs_test))
meta_model = LogisticRegression()
meta_model.fit(stacked_test, y_test)  # Usually fit on OOF training preds
ensemble_probs = meta_model.predict_proba(stacked_test)[:,1]
ensemble_pred = (ensemble_probs > 0.5).astype(int)

# -------------------------------
# 7. Compute all metrics
# -------------------------------
def compute_metrics(y_true, y_prob):
    y_pred = (y_prob > 0.5).astype(int)
    auc = roc_auc_score(y_true, y_prob)
    acc = accuracy_score(y_true, y_pred)
    auc_pr = average_precision_score(y_true, y_prob)
    bacc = balanced_accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    pcc, _ = pearsonr(y_true, y_prob)
    return {"AUC-ROC": auc, "ACC": acc, "AUC-PR": auc_pr, "BACC": bacc,
            "Precision": precision, "Recall": recall, "F1": f1, "Kappa": kappa,
            "RMSE": rmse, "PCC": pcc}

metrics = compute_metrics(y_test, ensemble_probs)
print(metrics)
```
